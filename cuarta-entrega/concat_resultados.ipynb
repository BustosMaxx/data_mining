```python
DIR = "/content/drive/MyDrive/DM/2026/datos"
DIR = "/content/drive/MyDrive/datasets/properati"
DIR_SOLUCIONES = f"{DIR}/soluciones"
DIR_local = "../data/stage"


lista_de_soluciones = [
    f"{DIR_local}/solucion-e4v4-casas-v3.csv",
    f"{DIR_local}/solucion-e4v4-cocheras.csv",
    f"{DIR_local}/solucion-e4v4-deptos-v4.csv",
    # seguir listando aca los archivos a concatenar
]

dfs_parciales = []
for f in lista_de_soluciones:
    s = pd.read_csv(f, index_col='id')['price']
    dfs_parciales.append(s)
```

```python
s_total = pd.concat(dfs_parciales)##  default axis=0 top-to-bottom, combine side-by-side result = pd.concat([df1, df2], axis=1)
# Si creen que puede haber duplicados, prueben está linea (no debería haberlos)
#s_total = s_total[~s_total.index.duplicated(keep='first')]
s_total = s_total.sort_index()
```
