# 1. Carga y Preparación de Datos

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_score
```

```python
df = pd.read_csv('/content/Automobile_data.csv')
df.shape
```

```python
df.head()
```

```python
# Eliminamos registros donde price no está presente, ya que es nuestra variable objetivo
df_test = df[df["price"] == '?'].copy()
df = df[df["price"] != '?'].copy()
```

```python
# Transformemos los '?' en NaNs
df.replace('?', pd.NA, inplace=True)

# A las columnas que queramos imputar para despues, las pasamos a numericas y le imputamos cero
cols_numericas = [
    "normalized-losses",
    "horsepower",
    "price",
    "bore",
    "stroke",
    "peak-rpm",
]
df[cols_numericas] = df[cols_numericas].apply(pd.to_numeric, errors='coerce')
df['num-of-doors'] = df['num-of-doors'].map({'four': 4, 'two': 2, pd.NA: 4})
df['num-of-cylinders'] = df['num-of-cylinders'].map({
    'two': 2,
    'three': 3,
    'four': 4,
    'five': 5,
    'six': 6,
    'eight': 8,
    'twelve': 12,
})

df[cols_numericas] = df[cols_numericas].fillna(df.mean(numeric_only=True))
```

```python
def feature_importances(modelo, df, topn=10):
    importances = modelo.feature_importances_
    indices = np.argsort(importances)[-topn:]

    plt.barh(range(len(importances[-topn:])), importances[indices])
    plt.yticks(range(len(importances[-topn:])), df.columns[indices])
    plt.title("Feature Importances")
    plt.tight_layout()
    plt.show()
```

Se arma un diccionario para ir recopilando información de cada experimento:

```python
results = {
    "nombre modelo": [],
    "RMSE score": [],
    "shape": [],
}
```

# 2. Modelo Baseline

```python
# Primer proceso predictivo
X = df.copy()
y = X.pop('price')

# Nuestro baseline tendrá solo las columnas numéricas originales
nombre_modelo = "Baseline"

X = X.select_dtypes(include='number')
baseline = RandomForestRegressor(n_jobs=-1, random_state=42)
baseline_score = cross_val_score(baseline, X, y, cv=5, scoring='neg_root_mean_squared_error').mean()
baseline_score = -1 * baseline_score.mean()

# Reportes de resultados
print(f"RMSE {nombre_modelo} Score: {baseline_score:.2f}")
baseline.fit(X, y)
feature_importances(baseline, X)
results["nombre modelo"].append(f"{len(results['nombre modelo']) + 1}-{nombre_modelo}")
results["RMSE score"].append(baseline_score)
results["shape"].append(X.shape)
print(pd.DataFrame(results).sort_values("RMSE score"))
```

# 3. Codificación de Variables Categóricas


## 3.1 Label Encoding

```python
# Primer proceso predictivo
X = df.copy()
y = X.pop('price')

# Nuestro baseline tendrá solo las columnas numéricas originales
nombre_modelo = "Label Encoding"

# Código para label encoding acá
for colname in X.select_dtypes(include=["object", "string"]):
    X[colname], _ = X[colname].factorize()

X = X.select_dtypes(include='number')
model = RandomForestRegressor(n_jobs=-1, random_state=42)
model_score = cross_val_score(model, X, y, cv=5, scoring='neg_root_mean_squared_error').mean()
model_score = -1 * model_score.mean()

# Reportes de resultados
print(f"RMSE {nombre_modelo} Score: {model_score:.2f}")
model.fit(X, y)
feature_importances(model, X)
results["nombre modelo"].append(f"{len(results['nombre modelo']) + 1}-{nombre_modelo}")
results["RMSE score"].append(model_score)
results["shape"].append(X.shape)
print(pd.DataFrame(results).sort_values("RMSE score"))
```

## 3.2 One-Hot Encoding

```python
# Primer proceso predictivo
X = df.copy()
y = X.pop('price')

# Nuestro baseline tendrá solo las columnas numéricas originales
nombre_modelo = "One hot Encoding"

# Código para label encoding acá
colnames = X.select_dtypes(include=["object", "string"]).columns
X = pd.get_dummies(X, columns=colnames, dummy_na=True, dtype=int)

X = X.select_dtypes(include='number')
model = RandomForestRegressor(n_jobs=-1, random_state=42)
model_score = cross_val_score(model, X, y, cv=5, scoring='neg_root_mean_squared_error').mean()
model_score = -1 * model_score.mean()

# Reportes de resultados
print(f"RMSE {nombre_modelo} Score: {model_score:.2f}")
model.fit(X, y)
feature_importances(model, X)
results["nombre modelo"].append(f"{len(results['nombre modelo']) + 1}-{nombre_modelo}")
results["RMSE score"].append(model_score)
results["shape"].append(X.shape)
print(pd.DataFrame(results).sort_values("RMSE score"))
```

```python
# Primer proceso predictivo
X = df.copy()
y = X.pop('price')

# Nuestro baseline tendrá solo las columnas numéricas originales
nombre_modelo = "Target Encoding"

# Target encoding
cols_categoricas = ['make', 'fuel-type', 'aspiration', 'body-style', 'drive-wheels',
                    'engine-location', 'engine-type', 'fuel-system']
X = X[X.select_dtypes(include='number').columns.tolist() + cols_categoricas].copy()
for col in cols_categoricas:
    media_por_cat = df.groupby(col)['price'].transform('mean')
    X[col + '_te'] = media_por_cat
    X.drop(columns=col, inplace=True)

X = X.select_dtypes(include='number')
model = RandomForestRegressor(n_jobs=-1, random_state=42)
model_score = cross_val_score(model, X, y, cv=5, scoring='neg_root_mean_squared_error').mean()
model_score = -1 * model_score.mean()

# Reportes de resultados
print(f"RMSE {nombre_modelo} Score: {model_score:.2f}")
model.fit(X, y)
feature_importances(model, X)
results["nombre modelo"].append(f"{len(results['nombre modelo']) + 1}-{nombre_modelo}")
results["RMSE score"].append(model_score)
results["shape"].append(X.shape)
print(pd.DataFrame(results).sort_values("RMSE score"))
```

# 4. Columnas matematicas derivadas

```python
# Primer proceso predictivo
X = df.copy()
y = X.pop('price')

# Nuestro baseline tendrá solo las columnas numéricas originales
nombre_modelo = "Cols Matematicas"

X["volumen"] = X["length"] * X["width"] * X["height"]
X["relacion_peso_potencia"] = X["horsepower"] / X["curb-weight"]
X["eficiencia_motor"] = X["horsepower"] / X["engine-size"]
X["precio_por_hp"] = y / X["horsepower"] # Peligroso!

X = X.select_dtypes(include='number')
model = RandomForestRegressor(n_jobs=-1, random_state=42)
model_score = cross_val_score(model, X, y, cv=5, scoring='neg_root_mean_squared_error').mean()
model_score = -1 * model_score.mean()

# Reportes de resultados
print(f"RMSE {nombre_modelo} Score: {model_score:.2f}")
model.fit(X, y)
feature_importances(model, X)
results["nombre modelo"].append(f"{len(results['nombre modelo']) + 1}-{nombre_modelo}")
results["RMSE score"].append(model_score)
results["shape"].append(X.shape)
print(pd.DataFrame(results).sort_values("RMSE score"))
```

# Normalización Z-Score

```python
# Primer proceso predictivo
X = df.copy()
y = X.pop('price')

# Nuestro baseline tendrá solo las columnas numéricas originales
nombre_modelo = "Normalización Z-Score"

numeric_cols = X.select_dtypes(include='number').columns
X[numeric_cols] = (X[numeric_cols] - X[numeric_cols].mean()) / X[numeric_cols].std()

X = X.select_dtypes(include='number')
model = RandomForestRegressor(n_jobs=-1, random_state=42)
model_score = cross_val_score(model, X, y, cv=5, scoring='neg_root_mean_squared_error').mean()
model_score = -1 * model_score.mean()

# Reportes de resultados
print(f"RMSE {nombre_modelo} Score: {model_score:.2f}")
model.fit(X, y)
feature_importances(model, X)
results["nombre modelo"].append(f"{len(results['nombre modelo']) + 1}-{nombre_modelo}")
results["RMSE score"].append(model_score)
results["shape"].append(X.shape)
print(pd.DataFrame(results).sort_values("RMSE score"))
```

# Transformaciones

```python
# Primer proceso predictivo
X = df.copy()
y = X.pop('price')

# Nuestro baseline tendrá solo las columnas numéricas originales
nombre_modelo = "Transformación Logaritmo"

for col in X.select_dtypes(include='number').columns:
    X[col] = np.log(abs(X[col])+1)

X = X.select_dtypes(include='number')
model = RandomForestRegressor(n_jobs=-1, random_state=42)
model_score = cross_val_score(model, X, y, cv=5, scoring='neg_root_mean_squared_error').mean()
model_score = -1 * model_score.mean()

# Reportes de resultados
print(f"RMSE {nombre_modelo} Score: {model_score:.2f}")
model.fit(X, y)
feature_importances(model, X)
results["nombre modelo"].append(f"{len(results['nombre modelo']) + 1}-{nombre_modelo}")
results["RMSE score"].append(model_score)
results["shape"].append(X.shape)
print(pd.DataFrame(results).sort_values("RMSE score"))
```

# Binning por K-Means

```python
# Primer proceso predictivo
X = df.copy()
y = X.pop('price')

# Nuestro baseline tendrá solo las columnas numéricas originales
nombre_modelo = "Binning por similaridad (kmeans)"

from sklearn.preprocessing import KBinsDiscretizer

n_bins = 5
numeric_cols = X.select_dtypes(include='number').columns
for col in numeric_cols:
    n_unique = X[col].nunique()
    reales_bins = min(n_bins, n_unique)
    discretizer = KBinsDiscretizer(n_bins=reales_bins, encode='ordinal', strategy='kmeans')
    X[col] = discretizer.fit_transform(X[[col]])


X = X.select_dtypes(include='number')
model = RandomForestRegressor(n_jobs=-1, random_state=42)
model_score = cross_val_score(model, X, y, cv=5, scoring='neg_root_mean_squared_error').mean()
model_score = -1 * model_score.mean()

# Reportes de resultados
print(f"RMSE {nombre_modelo} Score: {model_score:.2f}")
model.fit(X, y)
feature_importances(model, X)
results["nombre modelo"].append(f"{len(results['nombre modelo']) + 1}-{nombre_modelo}")
results["RMSE score"].append(model_score)
results["shape"].append(X.shape)
print(pd.DataFrame(results).sort_values("RMSE score"))
```

# Binning por entropia

```python
# Primer proceso predictivo
X = df.copy()
y = X.pop('price')

# Nuestro baseline tendrá solo las columnas numéricas originales
nombre_modelo = "Bining con Entropia"

from sklearn.tree import DecisionTreeRegressor

n_bins = 5
numeric_cols = X.select_dtypes(include='number').columns
for col in numeric_cols:
    tree = DecisionTreeRegressor(max_leaf_nodes=n_bins, random_state=42)
    tree.fit(X[[col]], y)
    X[col] = tree.apply(X[[col]])

X = X.select_dtypes(include='number')
model = RandomForestRegressor(n_jobs=-1, random_state=42)
model_score = cross_val_score(model, X, y, cv=5, scoring='neg_root_mean_squared_error').mean()
model_score = -1 * model_score.mean()

# Reportes de resultados
print(f"RMSE {nombre_modelo} Score: {model_score:.2f}")
model.fit(X, y)
feature_importances(model, X)
results["nombre modelo"].append(f"{len(results['nombre modelo']) + 1}-{nombre_modelo}")
results["RMSE score"].append(model_score)
results["shape"].append(X.shape)
print(pd.DataFrame(results).sort_values("RMSE score"))
```

# WoE

```python
# Primer proceso predictivo
X = df.copy()
y = X.pop('price')

# Nuestro baseline tendrá solo las columnas numéricas originales
nombre_modelo = "WoE"

numeric_cols = X.select_dtypes(include='number').columns

for col in numeric_cols:
    # armar un target binario
    target_binario = (y > y.median()).astype(int)

    bins = pd.qcut(X[col], q=n_bins, duplicates='drop')
    woe_df = pd.DataFrame({'bin': bins, 'target': target_binario})

    stats = woe_df.groupby('bin', observed=True)['target'].agg(['sum', 'count'])
    stats['no_evento'] = stats["count"] - stats["sum"]

    prob_evento = stats['sum'] / stats['sum'].sum()
    prob_no_evento = stats['no_evento'] / stats['no_evento'].sum()
    stats['woe'] = np.log((prob_evento + 1e-9) / (prob_no_evento + 1e-9))

    X[col] = bins.map(stats['woe']).astype('float')

X = X.select_dtypes(include='number')
model = RandomForestRegressor(n_jobs=-1, random_state=42)
model_score = cross_val_score(model, X, y, cv=5, scoring='neg_root_mean_squared_error').mean()
model_score = -1 * model_score.mean()

# Reportes de resultados
print(f"RMSE {nombre_modelo} Score: {model_score:.2f}")
model.fit(X, y)
feature_importances(model, X)
results["nombre modelo"].append(f"{len(results['nombre modelo']) + 1}-{nombre_modelo}")
results["RMSE score"].append(model_score)
results["shape"].append(X.shape)
print(pd.DataFrame(results).sort_values("RMSE score"))
```
